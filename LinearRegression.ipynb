# Linear Regression

#### X---> Features or Independent variables ---> Hourse studied
#### y---> Dependent Variable ---> Marks

```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
X=np.array([[1],[2],[3],[4],[5],[6]])
y=np.array([30,40,50,60,70,80])
```

####  Model Building

```python
Model=LinearRegression()
```

#### Model Training

```python
Model.fit(X,y)
```

#### Coefficient ---> m of y=mx+b

```python
Model.coef_
```

#### Intercept ---> b of y=mx+b

```python
Model.intercept_
```

#### Check if I study 7 hours how much marks I get

```python
result=Model.predict([[7]])
print(result)
result
```

### Multiple Linear Regression

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
X=pd.DataFrame({"Area":[20,25,30,35,40],"bedrooms":[1,2,3,4,5],"age":[20,22,24,26,28]})
y=np.array([10000,12000,14000,16000,18000])
```

```python
Model=LinearRegression()
```

```python
Model.fit(X,y)
```

```python
Model.coef_
```

```python
Model.intercept_
```

```python
result=Model.predict([[55,7,32]])
result
```

```python
new_data = pd.DataFrame({
    "Area": [55],
    "bedrooms": [7],
    "age": [32]
})

result = Model.predict(new_data)

print(result)
```
